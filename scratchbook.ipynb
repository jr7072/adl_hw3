```python
import json
import numpy as np

with open("./data/valid.json" , 'rb') as f:

    data = json.load(f)


digit_length = list()
for question, answer in data:

    answer_str = str(answer)
    digits = answer_str[answer_str.find(".") + 1:]
    len_digits = len(digits)

    digit_length.append(len_digits)

print(np.array(digit_length).mean())
print(np.median(np.array(digit_length)), "med")
print(np.quantile(np.array(digit_length), .9), "90th")
print(np.quantile(np.array(digit_length), .60), "60th")
print(np.quantile(np.array(digit_length), .75), "75th")
print(np.quantile(np.array(digit_length), .25), "25th")
print(np.quantile(np.array(digit_length), .99), "99th")
```
